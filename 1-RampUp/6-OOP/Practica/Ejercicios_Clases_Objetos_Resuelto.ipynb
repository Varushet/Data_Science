## Ejercicio 1
Somos una empresa que tiene varias tiendas de electrodomésticos. Necesitamos un programa para manejar las tiendas, ver las ventas que han tenido, cuántos empleados hay, etc... Para ello vamos a modelizar la tienda en una clase, que tendrá las siguientes características:
* Nombre clase: "Tienda"
* Atributos comunes:
    * Tipo: "Electrodomésticos"
    * Abierta: True
* Atributos propios de cada tienda:
    * Nombre - String
    * Dirección - String
    * Número de empleados - int
    * Ventas últimos 3 meses - Lista de 3 numeros
* Método para calcular las ventas de todos los meses, que devuelva un número con todas las ventas.
* Método que calcula la media de ventas de los últimos meses, por empleado
* Método que devuelve en un string el nombre de la tienda, junto con su dirección.
* Método que obtiene las ventas del último mes.
* Método para dar una proyección de las ventas en caso de haber invertido X dinero en marketing en esa tienda. Siendo X el parámetro de entrada. Si X es menor de 1000, las ventas de los ultimos 3 meses hubiesen sido de (1.2 x ventas), si es mayor o igual de 1000, las ventas hubiesen sido de (1.5 x venta). El método reescribe el atributo de ventas en los últimos 3 meses, y además devuelve las ventas con el aumento.

Se pide:
1. Implementar la clase "Tienda"
2. Crear tres tiendas con datos inventados
3. Comprueba en al menos una de ellas, todo lo implementado en la clase tienda (sus atributos, media de ventas, ventas/empleado...)
3. Calcular las ventas del último mes de todas las tiendas. Para ello usa el bucle `for`
4. Imprime por pantalla los nombres de las tiendas cuya dirección lleve el string "Avenida"
5. Seguro que ya has hecho este apartado... Documenta la clase :)

```python
# crea la clase con los atr fijos y el constructor (__init__) con los variables
class Tienda:
    tipo = "electrodomesticos"
    abierta = True
    def __init__(self, nombre, direccion, numero_de_empleados, ventas_tres_meses = [0, 0, 0]):
        self.nombre = nombre
        self.direccion = direccion
        self.numero_de_empleados = numero_de_empleados
        self.ventas_tres_meses = ventas_tres_meses

# el primero método() suma todas las venas y las guarda en una variable que pasa
    def calcular_ventas_trimestrales(self):
        total = sum(self.ventas_tres_meses)
        return total

# aprovechamos el resultado del primer método() para hacer una media sencilla y pasarla
    def media_ventas_empleado(self):
        total_ventas = self.calcular_ventas_trimestrales()
        media = total_ventas / self.numero_de_empleados
        return media

# unimos en un único string los atr solicitados
    def localizacion(self):
        return str(self.nombre + ": " + self.direccion)

# tomamos solo el último dato de la lista de ventas de los últimos tres meses, así aunque se añadan más datos, 
# el ultimo será siempre el más actualizado
    def ventas_recientes(self):
        return self.ventas_tres_meses[-1]

# sumamos las ventas en una variable total y la pasamos por la inversion
    def proyeccion(self, inversion):
        total = sum(self.ventas_tres_meses)
        if inversion < 1000:
            multi = 1.2
        else:
            multi = 1.5

# hacemos la proyeccion de ventas tras la inversion para cada tienda con un for-in
        nuevas_ventas = [venta * multi  for venta in self.ventas_tres_meses]

        return sum(nuevas_ventas)

    
```

```python
tienda_madrid = Tienda(
    nombre="ElectroMadrid Centro",
    direccion="Avenida las palmas 28, Madrid",
    numero_de_empleados=15,
    ventas_tres_meses=[12000, 14500, 13000] 
)
tienda_valencia = Tienda(
    nombre="ElectroPlaya Valencia",
    direccion="Calle de la Paz 14, Valencia",
    numero_de_empleados=6,
    ventas_tres_meses=[5000, 8200, 6100]
)
tienda_pueblo = Tienda(
    nombre="ElectroRural Soria",
    direccion="Plaza Mayor 1, Soria",
    numero_de_empleados=2,
    ventas_tres_meses=[1200, 950, 1100]
)

print(tienda_madrid.calcular_ventas_trimestrales())
print(tienda_madrid.media_ventas_empleado())
print(tienda_madrid.localizacion())
print(tienda_madrid.ventas_recientes())
print(tienda_madrid.proyeccion(500))
```

```python
#lista de objetos 
todas_tiendas = [tienda_madrid, tienda_valencia, tienda_pueblo]
total_general = 0

for tienda in todas_tiendas:
    ventas_ultimo_mes = tienda.ventas_recientes()
    total_general += ventas_ultimo_mes
    
print(f"VENTAS TOTALES (último mes) de todas las tiendas: {total_general}€")

# bandera para controlar la salida
tienda_avenida = False

# busca y normaliza en cada objeto el atr solicitado y lo pasa por el comprobante
for tienda in todas_tiendas:
    if "avenida" in tienda.direccion.lower():
        print(tienda.nombre + ": " + tienda.direccion)
        tienda_avenida = True
if not tienda_avenida:
        print("No hay tiendas en avenidas")
```

## Ejercicio 2
En este ejercicio vamos a implementar una clase *Perro* en Python. La clase tiene las siguientes características:
* Cosas que sabemos seguro que tiene un perro
    * Tiene 4 patas
    * 2 orejas
    * 2 ojos
    * Una velocidad de 0. Por defecto, el perro está parado
* Cuando se inicialice:
    * El perro será de una determinada raza
    * Por defecto tendrá pelo "Marrón", a no ser que se diga lo contrario.
    * Por defecto no tendrá dueño, a no ser que se diga lo contrario.
    
* Dispondrá también de un método llamado andar, que tiene un argumento de entrada (aumento_velocidad). Este valor se le sumará a la velocidad que ya llevaba el perro.
* Necesita otro método (parar), donde pondremos la velocidad a 0.
* Otro método llamado "ladrar", que tendrá un argumento de entrada, y la salida será el siguiente string: "GUAU!" + el argumento de entrada.


Se pide:
* Implementa la clase *Perro*
* Crea un objeto de tipo *Perro*, sin dueño
* Comprueba que están bien todos sus atributos
* Prueba que ande, y comprueba su velocidad
* Páralo
* Documenta la clase *Perro*

```python
# Creamos la clase y us atr fijos
class Perro:
    patas = 4
    orejas = 2
    ojos = 2
    velocidad = 0
    
#Con el constructor atribuimos los argumentos
    def __init__ (self, nombre, raza, pelo = "marron", duenno = False):
        self.nombre = nombre
        self.raza = raza
        self.pelo = pelo
        self.duenno = duenno
    
#Un primer metodo() para cambiar la velocidad
    def andar(self, aumento_vel):
        self.velocidad += aumento_vel
        #return self.velocidad
        print(f"{self.nombre} se puso en marcha a {self.velocidad} Km/h")

#Reseteamos el atributo a 0 para "detenerlo"
    def parar(self):
        self.velocidad = 0
        print(f"{self.nombre} se detuvo: {self.velocidad} Km/h")

#sumamos el argumento con un injerto solicitado
    def ladrar(self, entrada = None):
        if entrada == None: # <-- Un Fallback
            entrada = input("A qué quieres que le ladre?")
        return entrada + ", " + "GUAU!"
        
Arturo = Perro("Arturo", "Pug")

Arturo.andar(6)
Arturo.parar()
print(Arturo.ladrar())

print(Arturo.__dict__, "Patas:", Arturo.patas, "Orejas:", Arturo.orejas, "Ojos:", Arturo.ojos, "Velocidad:", Arturo.velocidad)
```
