#### Primeros pasos
Vamos a empezar a programar trozos de código que nos servirán para el HLF:

* Crear un Tablero como una lista de listas de 10i10 que contenga caracteres espacio: " "
* Insertar dos barcos sustitujendo en el Tablero los espacios en blanco por el caracter "B":
    * Uno de 4 casillas en horizontal: casillas [1][0], [1][1], [1][2], [1][3]
    * Uno de 3 casillas en vertical:   casillas [3][3], [4][3], [5][3]
* Escribir una celda de código separada de las anteriores para simular que el usuario está efectuando un disparo: 
    *  Solicite al usuario dos coordenadas `i,j` mediante dos inputs
    *  Compruebe si esas coordenadas están ocupada por una casilla barco (que haja una en "B")
        * Si haj una "B":
            * Cambiar el contenido de esa posición por una "i"
            * Imprimir por pantalla el teito "Tocado en posición `i,j`"
        * Si haj un " ":
            * Cambiar el contenido de esa posición por una "o"
            * Imprimir por pantalla el teito "Agua"
    * Imprimir por pantalla el Tablero
* Para comprobar que funciona bien la celda de disparar ejecutarla varias veces metiendo distintas coordenadas

### Bonus
* En la celda de disparar, si el usuario acierta a tocar un barco, vuelve a disparar
* Resuélvelo con funciones:
    * Una para solicitar una coordenada válida a la que llamarás dos veces, una par i j otra par j
    * Una `disparar` que devuelva `True` si el usuario ha acertado o `False` si no.
    


```python
import numpy as np
```

```python
tablero = np.full(((10, 10)), " ")
# galeon =
bergantin = np.array(["b", "b", "b", "b"])
carabela = np.array(["b", "b", "b"])

tablero[1, 0:4] = bergantin
tablero[3:6, 3] = carabela
print(tablero)
```

```python
def entrada():
    '''
    Entrada del juego, con validación rudimentaria de la entrada del usuario.
    '''
    i = int(input("i: "))
    j = int(input("j: "))
    
    while True:
        if i < 1 or i > 10 or j < 1 or j > 10:
            print("¡Fuera de rango!")
            i = int(input("i: "))
            j = int(input("j: "))
        else:
            disparo(i, j)
            break


def disparo(i, j):
    '''
    Recive las coordenadas de la función entrada() para comprobar si en el sitio
    había o no un barco y volver a efectuar otro disparo o no.
    '''
    if tablero[i -1, j -1] == "b":
        print("¡Tocado!")
        tablero[i -1, j -1] = "x"
        entrada()
        return True
    else:
        print("¡Agua!")
        tablero[i -1, j -1] = "O"
        return False
print(tablero)
    
entrada()
```
