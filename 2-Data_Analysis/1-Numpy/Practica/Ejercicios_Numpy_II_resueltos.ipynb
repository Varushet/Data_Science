## Ejercicio 12
Encuentra los índices de los valores que no sean 0 del siguiente vector [1,2,0,0,4,0]. El output sería [1, 2 4]. [Encuentra el método apropiado en la documentación.](https://numpy.org/doc/) Encapsúlalo en una función, con un argumento de entrada de un vector.

````python
victor = np.array([1, 2, 0, 0, 4, 0])

def encuentra_cero(vector):
    mascara = np.array(vector != 0)
    return vector[mascara]
    # return np.nonzero(vector)[0]

encuentra_cero(victor)

#El output propuesto está mal, son los valores, no los índices
````

````python
def encuentra_no_cero(vector):
    # nonzero() devuelve una tupla de arrays, tomamos el primer elemento [0]
    # que contiene los índices de los elementos no nulos
    return np.nonzero(vector)[0]

encuentra_no_cero(victor)
````

## Ejercicio 18
Crea una matriz aleatoria de 5x5 y normaliza sus valores. Para normalizar tienes que restar sus valores por la media(`np.mean`) y dividir el resultado por la desviación estándar(`np.std`).

````python
quintal = np.random.randint(25, size=100)
print(quintal)
media = np.mean(quintal)
desviacion = np.std(quintal)
normalizado = (quintal - media) / desviacion
print(normalizado)
````
