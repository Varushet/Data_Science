### Step 1. Import the necessary libraries

```python
import pandas as pd
import numpy as np
```

### Step 2. Import the dataset from this [address](https://raw.githubusercontent.com/justmarkham/DAT8/master/data/chipotle.tsv) and assign it to a variable called chipo.

```python
url = "https://raw.githubusercontent.com/justmarkham/DAT8/master/data/chipotle.tsv"
chipo = pd.read_csv(url, sep="\t")
chipo
```

### Step 3. Name of the max valued product

```python
chipo.loc[chipo['item_price'].idxmax(), 'item_name']
# chipo.loc[(chipo['item_price']/chipo["quantity"]).idxmax(), 'item_name']
```

### Step 4. How many products cost more than $10.00?

```python
chipo['item_price_num'] = chipo['item_price'].str.replace('$', '').astype(float)
chipo[chipo['item_price_num'] > 10]['item_name'].unique()
```

### Step 4.1: Y cuántos pedidos se han hecho con un producto de más de 10$? Es lo mismo?

```python
len(chipo[chipo['item_price_num'] > 10]['order_id'].unique())
```

### Step 4.2: Y cuántos pedidos se han hecho de más de 10$? Es lo mismo?

```python
# chipo.groupby('order_id')['item_price_num'].apply(list)
order_price = chipo.groupby('order_id')['item_price_num'].sum()

print(len(order_price > 10))

len(chipo[chipo['item_price_num'] > 10]['order_id']) == len(order_price > 10)
```

### Step 4.3: Y en cuántos pedidos se ha pagado más de 10$ por un mismo producto? Es lo mismo?

```python
pedidos_mayores_diez = chipo[chipo['item_price_num'] > 10]['order_id'].nunique()
print(pedidos_mayores_diez)

print(len(order_price > 10) == pedidos_mayores_diez)
```

### Step 5. What is the price of each item and name it unit_price. Get only item_name and unit_price

```python
unit_price = chipo[['item_name', 'item_price']].drop_duplicates().sort_values('item_name')
print(unit_price)
```
