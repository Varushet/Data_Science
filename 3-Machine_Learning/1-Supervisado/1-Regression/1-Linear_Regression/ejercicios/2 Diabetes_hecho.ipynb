## Building a regression model

The data have already been split to predictor (*data*) and response (*target*) variables. Given this information, we'll apply what we have previously learned about linear regression and apply the algorithm to the diabetes dataset.

#### In the cell below, import the `linear_model` class from `sklearn`. 

```python
from sklearn import linear_model
```

#### Create a new instance of the linear regression model and assign the new instance to the variable `diabetes_model`.

```python
diabetes_model = linear_model.LinearRegression()
```

#### Next, let's split the training and test data.

Define `diabetes_data_train`, `diabetes_target_train`, `diabetes_data_test`, and `diabetes_target_test`. Use the last 20 records for the test data and the rest for the training data.

```python
n_samples = diabetes['data'].shape[0]
n_test = 20

diabetes_data_train = diabetes['data'][:-n_test]
diabetes_target_train = diabetes['target'][:-n_test]

diabetes_data_test = diabetes['data'][-n_test:]
diabetes_target_test = diabetes['target'][-n_test:]

print(f"Datos de entrenamiento: {diabetes_data_train.shape}")
print(f"Datos de prueba: {diabetes_data_test.shape}")
```

Perform a brief exploratory analysis to investigate the conditions necessary for linear regression. This analysis will help you understand the distribution of the target variable and its correlations with other features.

Fit the training data and target to `diabetes_model`. Print the *intercept* and *coefficients* of the model.

```python
diabetes_model.fit(diabetes_data_train, diabetes_target_train)
```

```python
print("Intercepto:", diabetes_model.intercept_)

print("Coeficientes:", diabetes_model.coef_)
```

#### Inspecting the results

From the outputs you should have seen:

- The intercept is a float number.
- The coefficients are an array containing 10 float numbers.

This is the linear regression model fitted to your training dataset.

#### Using your fitted linear regression model, predict the *y* of `diabetes_data_test`.

```python
diabetes_predictions = diabetes_model.predict(diabetes_data_test)
diabetes_predictions[:5]
```

#### Print your `diabetes_target_test` and compare with the prediction. 

```python
comparison_df = pd.DataFrame({
    'Valor Real': diabetes_target_test,
    'Predicción': diabetes_predictions
})
comparison_df
```

#### Is `diabetes_target_test` exactly the same as the model prediction? Use different error metrics

```python
from sklearn.metrics import mean_squared_error, r2_score
```

```python
mse = mean_squared_error(diabetes_target_test, diabetes_predictions)
mse
```

#### Iterate with a new model to assess if any improvements can be made compared to the previous one. Consider reducing the number of variables or using `StandardScaler` to enhance model performance.

```python
from sklearn.preprocessing import StandardScaler
from sklearn.pipeline import Pipeline
```

```python
pipeline_model = Pipeline([
    ('scaler', StandardScaler()),
    ('regression', linear_model.LinearRegression())
])
```

```python
# 2. Entrenar el pipeline con los datos de train
pipeline_model.fit(diabetes_data_train, diabetes_target_train)
```

```python
# 3. Predecir con test
pipeline_predictions = pipeline_model.predict(diabetes_data_test)
```

```python
# 4. Evaluar
mse_new = mean_squared_error(diabetes_target_test, pipeline_predictions)
r2_new = r2_score(diabetes_target_test, pipeline_predictions)
```

```python
print(mse_new, r2_new)
```
